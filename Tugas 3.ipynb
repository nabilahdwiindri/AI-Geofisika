### 2. Nyatakan Minimal Theory dan Assumptions

Transformasi Fourier merupakan metode matematika untuk merepresentasikan sinyal sebagai kombinasi komponen frekuensi. Sinyal yang awalnya dinyatakan sebagai fungsi waktu dapat dianalisis berdasarkan kontribusi setiap frekuensi penyusunnya.

#### 2.1. Persamaan dasar

Untuk sinyal kontinu, Transformasi Fourier dapat ditulis sebagai:

\[
X(f)=\int_{-\infty}^{\infty}
x(t)e^{-i2\pi ft}\,dt
\]

Keterangan:

- \(x(t)\): sinyal dalam domain waktu.
- \(X(f)\): representasi sinyal dalam domain frekuensi.
- \(t\): waktu dalam detik.
- \(f\): frekuensi dalam hertz (Hz).
- \(i\): unit imajiner.

Untuk data digital, digunakan Discrete Fourier Transform (DFT):

\[
X[k]=\sum_{n=0}^{N-1}
x[n]e^{-i2\pi kn/N}
\]

Keterangan:

- \(x[n]\): sampel sinyal ke-\(n\).
- \(N\): jumlah sampel.
- \(k\): indeks frekuensi.
- \(X[k]\): koefisien kompleks frekuensi ke-\(k\).

Perhitungan DFT dilakukan menggunakan algoritma FFT agar lebih efisien.

#### 2.2. Frekuensi sampling dan resolusi frekuensi

Jika interval sampling adalah \(\Delta t\), frekuensi sampling dihitung dengan:

\[
f_s=\frac{1}{\Delta t}
\]

Frekuensi Nyquist adalah:

\[
f_N=\frac{f_s}{2}
\]

Frekuensi Nyquist merupakan batas frekuensi tertinggi yang dapat direpresentasikan tanpa aliasing dalam kondisi sampling yang memenuhi teorema Nyquist.

Resolusi frekuensi untuk rekaman berdurasi \(T\) adalah:

\[
\Delta f=\frac{f_s}{N}=\frac{1}{T}
\]

Semakin panjang durasi rekaman, semakin kecil jarak antarbin frekuensi. Namun, kemampuan memisahkan frekuensi yang berdekatan juga dipengaruhi oleh bentuk sinyal dan pemilihan window.

#### 2.3. Spektrum amplitudo satu sisi

Untuk sinyal real, spektrum amplitudo satu sisi dapat dihitung dari FFT dengan normalisasi:

\[
A[k]=\frac{|X[k]|}{N}
\]

Nilai amplitudo pada bin frekuensi positif selain DC dan, jika \(N\) genap, bin Nyquist dikalikan dua. Hal ini mempertahankan amplitudo sinusoidal dalam satuan asli sinyal.

Frekuensi dominan ditentukan dengan:

\[
f_{\mathrm{dom}}=
\underset{f_k>0}{\operatorname{argmax}}\,A[k]
\]

Artinya, frekuensi dominan merupakan frekuensi nonnol dengan amplitudo terbesar pada spektrum yang dihitung.

#### 2.4. Asumsi

1. Data sinyal berupa sampel numerik dengan interval sampling yang seragam.
2. Waktu pengukuran dinyatakan dalam detik.
3. Frekuensi dinyatakan dalam Hz.
4. Amplitudo menggunakan satuan sesuai instrumen, misalnya m/s, m/s², atau satuan digital.
5. Data yang dianalisis dianggap mewakili rekaman yang akan dikaji.
6. Tidak terdapat nilai kosong atau nilai nonnumerik pada kolom utama.
7. Frekuensi dominan dihitung setelah pengurangan nilai rata-rata sinyal.
8. Tidak ada klaim bahwa frekuensi di atas Nyquist dapat dipulihkan dari data yang sudah mengalami aliasing.

Pengurangan nilai rata-rata menghilangkan komponen DC, tetapi tidak menghilangkan seluruh noise atau tren sinyal. Untuk sinyal nonstasioner, analisis FFT keseluruhan rekaman dapat menyembunyikan perubahan frekuensi terhadap waktu.

### 5. Implementasikan Procedure sebagai Artifact

Program berikut disimpan sebagai `fourier_analysis.py`.

Program menggunakan FFT untuk menghitung spektrum amplitudo satu sisi, memilih frekuensi dominan nonnol, dan menghasilkan grafik serta laporan.


````python
from pathlib import Path
import hashlib

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

SOURCE = Path("raw/seismic_signal.csv")
OUTPUT = Path("output")

TIME_COL = "time"
SIGNAL_COL = "amplitude"

def sha256(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()

def main():
    if not SOURCE.exists():
        raise FileNotFoundError(
            f"Input tidak ditemukan: {SOURCE}"
        )

    OUTPUT.mkdir(parents=True, exist_ok=True)
    source_hash_before = sha256(SOURCE)

    df = pd.read_csv(SOURCE)

    required = [TIME_COL, SIGNAL_COL]
    missing = [c for c in required if c not in df.columns]

    if missing:
        raise ValueError(
            f"Kolom wajib tidak tersedia: {missing}"
        )

    df = df[required].copy()

    for col in required:
        df[col] = pd.to_numeric(
            df[col], errors="coerce"
        )

    if df.isna().any().any():
        raise ValueError(
            "Data mengandung nilai kosong atau nonnumerik."
        )

    time = df[TIME_COL].to_numpy(dtype=float)
    signal = df[SIGNAL_COL].to_numpy(dtype=float)

    if len(time) < 4:
        raise ValueError(
            "Diperlukan minimal empat sampel."
        )

    if not np.isfinite(time).all():
        raise ValueError("Data waktu tidak finite.")

    if not np.isfinite(signal).all():
        raise ValueError("Sinyal mengandung nilai non-finite.")

    dt_all = np.diff(time)

    if np.any(dt_all <= 0):
        raise ValueError(
            "Waktu harus meningkat secara ketat."
        )

    dt = float(np.median(dt_all))

    if not np.allclose(
        dt_all, dt, rtol=1e-4, atol=1e-10
    ):
        raise ValueError(
            "Interval sampling tidak seragam."
        )

    fs = 1.0 / dt
    nyquist = fs / 2.0
    n = len(signal)
    duration = (n - 1) * dt

    # Hilangkan komponen DC.
    centered = signal - np.mean(signal)

    # FFT untuk sinyal real.
    fft_values = np.fft.rfft(centered)
    frequencies = np.fft.rfftfreq(n, d=dt)

    # Spektrum amplitudo satu sisi.
    amplitude = np.abs(fft_values) / n

    if n % 2 == 0:
        amplitude[1:-1] *= 2
    else:
        amplitude[1:] *= 2

    # Abaikan frekuensi nol ketika mencari frekuensi dominan.
    if np.allclose(centered, 0):
        dominant_frequency = None
        dominant_amplitude = None
        status = "REVIEW"
    else:
        nonzero = frequencies > 0
        indices = np.flatnonzero(nonzero)
        peak_index = indices[
            np.argmax(amplitude[nonzero])
        ]

        dominant_frequency = float(
            frequencies[peak_index]
        )
        dominant_amplitude = float(
            amplitude[peak_index]
        )
        status = "PASS"

    spectrum = pd.DataFrame({
        "frequency_hz": frequencies,
        "amplitude": amplitude,
    })

    spectrum.to_csv(
        OUTPUT / "spectrum.csv",
        index=False,
    )

    # Grafik sinyal domain waktu.
    plt.figure(figsize=(10, 4))
    plt.plot(time, signal)
    plt.xlabel("Time (s)")
    plt.ylabel("Amplitude")
    plt.title("Geophysical Signal in Time Domain")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(
        OUTPUT / "time_signal.png", dpi=150
    )
    plt.close()

    # Grafik spektrum amplitudo.
    plt.figure(figsize=(10, 4))
    plt.plot(frequencies, amplitude)
    plt.xlabel("Frequency (Hz)")
    plt.ylabel("Amplitude")
    plt.title("Single-Sided Amplitude Spectrum")
    plt.xlim(0, nyquist)
    plt.grid(True)

    if dominant_frequency is not None:
        plt.axvline(
            dominant_frequency,
            linestyle="--",
            label=f"Peak: {dominant_frequency:.4f} Hz",
        )
        plt.legend()

    plt.tight_layout()
    plt.savefig(
        OUTPUT / "amplitude_spectrum.png", dpi=150
    )
    plt.close()

    source_hash_after = sha256(SOURCE)

    summary = [
        f"Status: {status}",
        f"Samples: {n}",
        f"Sampling interval (s): {dt}",
        f"Sampling frequency (Hz): {fs}",
        f"Nyquist frequency (Hz): {nyquist}",
        f"Frequency resolution (Hz): {fs / n}",
        f"Record duration (s): {duration}",
        f"Dominant frequency (Hz): {dominant_frequency}",
        f"Dominant amplitude: {dominant_amplitude}",
        f"Source SHA-256 before: {source_hash_before}",
        f"Source SHA-256 after: {source_hash_after}",
    ]

    (OUTPUT / "analysis_summary.txt").write_text(
        "\n".join(summary) + "\n",
        encoding="utf-8",
    )

    print("\n".join(summary))

if __name__ == "__main__":
    main()
````

**Catatan implementasi:**

- Program mengasumsikan interval sampling seragam. Data dengan sampling tidak seragam harus ditangani menggunakan metode yang sesuai sebelum FFT standar digunakan.
- Resolusi frekuensi dihitung menggunakan \(f_s/N\), sedangkan durasi dari sampel pertama hingga terakhir adalah \((N-1)\Delta t\).
- Pengurangan rata-rata menghilangkan komponen DC, tetapi tidak menghilangkan seluruh tren atau noise.
- Program belum menggunakan window seperti Hann. Untuk sinyal yang tidak mencakup jumlah siklus bulat, spectral leakage dapat memengaruhi spektrum. Window dapat ditambahkan jika diperlukan.
- Program ini menghasilkan spektrum amplitudo, bukan Power Spectral Density (PSD).
