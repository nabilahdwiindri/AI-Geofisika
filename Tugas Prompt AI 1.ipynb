# Output Prompt 1 — Respon Dosen Geomath (Turunan & Persamaan Diferensial)

---

### 1. Tujuan Pembelajaran
Setelah mempelajari materi ini, mahasiswa diharapkan dapat:
- Memahami konsep fisik turunan sebagai laju perubahan instan (kemiringan/gradien).
- Menguasai teknik turunan dasar, aturan rantai (*chain rule*), dan turunan orde tinggi.
- Memahami konsep dasar Persamaan Diferensial (PD) dan penyelesaiannya.
- Menerapkan konsep turunan dalam kasus nyata Geofisika (seperti gradien geotermal dan percepatan gelombang).

---

### 2. Konsep Dasar
Dalam Geofisika, besaran-besaran alam (seperti suhu bumi, tekanan, medan gravitasi, dan simpangan gelombang) selalu berubah terhadap ruang (jarak $x, y, z$) dan waktu ($t$). 
**Turunan** adalah alat matematika untuk mengukur **seberapa cepat suatu besaran berubah** saat variabel lain berubah sedikit sekali (mendekati nol).

---

### 3. Penjelasan Sederhana
Bayangkan kamu sedang mengendarai mobil di daerah gunung saat mengukur nilai anomali gravitasi:
- **Kecepatan rata-rata:** Jarak total dibagi waktu total ($\Delta x / \Delta t$).
- **Kecepatan sesaat (Spedometer):** Kecepatan pada **detik ini juga** ($\Delta t \to 0$). Inilah turunan pertama: $v(t) = \frac{dx}{dt}$.
- Jika kamu menekan pedal gas, kecepatan berubah terhadap waktu. Laju perubahan kecepatan inilah yang disebut **percepatan** (turunan kedua): $a(t) = \frac{dv}{dt} = \frac{d^2x}{dt^2}$.

---

### 4. Rumus Penting

1. **Turunan Fungsi Pangkat (Power Rule):**
   $$\frac{d}{dx}(a x^n) = a \cdot n x^{n-1}$$
   *Arti simbol:* $a$ = koefisien, $x$ = variabel independen, $n$ = pangkat (eksponen).

2. **Aturan Rantai (Chain Rule):**
   $$\frac{dy}{dx} = \frac{dy}{du} \cdot \frac{du}{dx}$$
   *Digunakan jika fungsi berada di dalam fungsi lain ($y = f(u)$ di mana $u = g(x)$).*

3. **Turunan Kedua (Second Derivative):**
   $$\frac{d^2y}{dx^2} = \frac{d}{dx}\left(\frac{dy}{dx}\right)$$
   *Arti simbol:* Turunan dari turunan pertama, menyatakan kelengkungan atau percepatan.

4. **Persamaan Diferensial Dasar (Bentuk Eksponensial):**
   $$\frac{dy}{dx} = k y \implies y(x) = C e^{kx}$$
   *Arti simbol:* $k$ = konstanta laju pertumbuhan/peluruhan, $C$ = kondisi awal ($y(0)$).

---

### 5. Contoh Soal Mudah
**Soal:** Hitung turunan pertama dan kedua dari fungsi posisi gelombang permukaan $f(x) = 4x^3 - 5x^2 + 7x - 9$.

**Pembahasan Langkah demi Langkah:**
1. *Turunan Pertama ($f'(x)$):*
   - Turunan $4x^3$ adalah $4 \cdot 3 x^{3-1} = 12x^2$.
   - Turunan $-5x^2$ adalah $-5 \cdot 2 x^{2-1} = -10x$.
   - Turunan $7x$ adalah $7 \cdot 1 x^{1-1} = 7$.
   - Turunan konstanta $-9$ adalah $0$.
   - Jadi, **$f'(x) = 12x^2 - 10x + 7$**.

2. *Turunan Kedua ($f''(x)$):*
   - Turunan $12x^2$ adalah $12 \cdot 2 x^{2-1} = 24x$.
   - Turunan $-10x$ adalah $-10$.
   - Turunan konstanta $7$ adalah $0$.
   - Jadi, **$f''(x) = 24x - 10$**.

---

### 6. Contoh Soal Tingkat Sedang (Aturan Rantai)
**Soal:** Tentukan turunan pertama dari fungsi $y = (2x^3 + 5)^4$.

**Pembahasan Langkah demi Langkah:**
1. Misalkan fungsi bagian dalam: $u = 2x^3 + 5$.
2. Maka fungsi utama menjadi: $y = u^4$.
3. Cari turunan masing-masing:
   - $\frac{dy}{du} = 4u^3$
   - $\frac{du}{dx} = 6x^2$
4. Terapkan Aturan Rantai ($\frac{dy}{dx} = \frac{dy}{du} \cdot \frac{du}{dx}$):
   $$\frac{dy}{dx} = (4u^3) \cdot (6x^2) = 24x^2 u^3$$
5. Kembalikan $u$ ke variabel aslinya:
   $$\frac{dy}{dx} = 24x^2 (2x^3 + 5)^3$$

---

### 7. Contoh Soal Tingkat Geofisika (Gradien Geotermal & Aliran Panas)
**Soal:** Profil suhu $T$ (dalam $^\circ\text{C}$) terhadap kedalaman $z$ (dalam $\text{km}$) di suatu wilayah eksplorasi geotermal dinyatakan oleh:
$$T(z) = 25 + 30z - 0.5z^2$$
1. Tentukan persamaan gradien suhu ($\frac{dT}{dz}$).
2. Hitung gradien suhu pada kedalaman $z = 4\text{ km}$.
3. Menurut Hukum Fourier, rapat arus konduksi panas adalah $q = -K \frac{dT}{dz}$. Jika konduktivitas termal batuan $K = 2.5\text{ W/m}^\circ\text{C}$, tentukan nilai $q$ pada kedalaman $z = 4\text{ km}$.

**Pembahasan Langkah demi Langkah:**
1. *Mencari Gradien Suhu ($\frac{dT}{dz}$):*
   $$\frac{dT}{dz} = \frac{d}{dz}(25) + \frac{d}{dz}(30z) - \frac{d}{dz}(0.5z^2) = 0 + 30 - 1.0z = 30 - z$$
2. *Hitung pada $z = 4\text{ km}$:*
   $$\frac{dT}{dz}\Big|_{z=4} = 30 - 4 = 26^\circ\text{C/km} = 0.026^\circ\text{C/m}$$
3. *Hitung Aliran Panas $q$:*
   $$q = -K \frac{dT}{dz} = -2.5 \times 0.026 = -0.065\text{ W/m}^2$$
   *(Tanda negatif menunjukkan panas mengalir dari tempat panas (dalam bumi) ke tempat dingin (permukaan)).*

# Output Prompt 2 — Respon Tutor Geofisika Komputasi (Pengolahan Medan Gravitasi & Magnet dengan MATLAB)

---

### 1. Analisis Struktur Data
Data anomali medan potensial (gravitasi/magnet) umumnya berformat matriks atau ASCII 2-kolom:
- **Kolom 1 ($x$):** Jarak/Posisi lintasan pengukuran (dalam satuan meter [m] atau kilometer [km]).
- **Kolom 2 ($y$ / $\Delta B$ / $\Delta g$):** Amplitudo anomali medan magnet (nT) atau gravitasi (mGal).
- **Syarat Data:** Data harus terurut secara spasial dengan jarak antar sampel yang seragam ($\Delta x = \text{konstan}$).

---

### 2. Penjelasan Metode FFT (Fast Fourier Transform)
FFT mengonversi sinyal anomali dari **Domain Spasial** (amplitudo vs posisi $x$) menjadi **Domain Frekuensi Spasial / Bilangan Gelombang** (power vs $k$, di mana $k = 1/\lambda$).
- **Frekuensi Tinggi (Panjang Gelombang Pendek):** Berasal dari sumber anomali dangkal atau *noise*.
- **Frekuensi Rendah (Panjang Gelombang Panjang):** Berasal dari struktur anomali dalam (*regional anomaly*).

---

### 3. Langkah Pengolahan
1. **Load & Preprocessing:** Membaca file data, menghapus *mean* (*demean*) dan tren linier (*detrend*) agar tidak terjadi efek kebocoran spektral (*spectral leakage*).
2. **Sampling Parameters:** Menghitung interval sampel $\Delta x$ dan Nyquist Wavenumber.
3. **Kalkulasi FFT:** Menghitung FFT 1D menggunakan fungsi `fft()` di MATLAB dan mengubah spektrum menjadi *Power Spectrum* ($|F(k)|^2$).
4. **Visualisasi & Interpretasi:** Plot *Power Spectrum* skala semi-logaritma ($\ln(P)$ vs $k$) untuk mengestimasi kedalaman rata-rata sumber anomali (metode Spector & Grant, 1970).

---

### 4. Kode MATLAB Lengkap

```matlab
% =========================================================================
% PROGRAM PENGOLAHAN DATA GEOFISIKA: ANALISIS SPEKTRAL FFT 1D
% Mata Kuliah: Geofisika Komputasi
% =========================================================================
clc; clear; close all;

%% 1. MEMBACA & MEMBUAT DATA SINTETIS GEOFISIKA
% Jika menggunakan file eksternal, gunakan: data = readmatrix('data_geofisika.txt');
% Di sini dibuat contoh sinyal anomali magnet (Kombinasi Sumber Dalam + Dangkal + Noise)
x = 0:10:2000; % Lintasan 0 - 2000 meter dengan sampling dx = 10 m
dx = x(2) - x(1); % Interval sampling (m)
N = length(x); % Jumlah sampel data

% Sintesa sinyal anomali (Regional + Residual + Noise)
sumber_dalam = 50 * exp(-((x - 1000)/300).^2);
sumber_dangkal = 15 * sin(2*pi*x/150);
noise = 2 * randn(size(x));
anomali = sumber_dalam + sumber_dangkal + noise;

%% 2. PREPROCESSING DATA
% Demean dan Detrend data untuk menghilangkan Offset DC & Tren Linier
anomali_detrend = detrend(anomali - mean(anomali));

%% 3. PERHITUNGAN FAST FOURIER TRANSFORM (FFT)
Y = fft(anomali_detrend);

% Menghitung Power Spectrum Dua Sisi & Satu Sisi
P2 = abs(Y/N).^2;
P1 = P2(1:floor(N/2)+1);
P1(2:end-1) = 2 * P1(2:end-1);

% Vektor Frekuensi Spasial (k = 1/lambda) dalam cycle/meter
k = (0:(N/2)) / (N * dx); 
ln_P = log(P1 + 1e-10); % Skala Natural Logarithm untuk Power Spectrum

%% 4. VISUALISASI GRAFIK
figure('Name', 'Analisis FFT Data Geofisika', 'NumberTitle', 'off');

% Subplot 1: Anomali Medan Spasial
subplot(2,1,1);
plot(x, anomali, 'b-', 'LineWidth', 1.5); hold on;
plot(x, anomali_detrend, 'r--', 'LineWidth', 1.2);
title('Profil Anomali Medan Magnet / Gravitasi Spasial');
xlabel('Jarak Posisi x (meter)');
ylabel('Amplitudo Anomali (nT / mGal)');
legend('Data Asli', 'Data Preprocessed (Detrend)');
grid on;

% Subplot 2: Spektrum Daya (Power Spectrum)
subplot(2,1,2);
plot(k, ln_P, 'k-o', 'LineWidth', 1.5, 'MarkerFaceColor', 'r', 'MarkerSize', 4);
title('Spektrum Daya Spasial (Power Spectrum FFT)');
xlabel('Frekuensi Spasial k (cycles/meter)');
ylabel('ln(Power Spectrum)');
grid on;
```

---

### 5. Penjelasan Kode Baris demi Baris
- `clc; clear; close all;`: Membersihkan layar *Command Window*, menghapus variabel memori, dan menutup semua jendela grafik sebelumnya.
- `dx = x(2) - x(1);`: Menentukan interval ruang antar titik ukur ($\Delta x$).
- `detrend(...)`: Menghapus komponen tren linier agar respon frekuensi $k=0$ (DC component) tidak mendominasi spektrum.
- `fft(...)`: Mengonversi matriks sinyal domain spasial menjadi deret kompleks domain frekuensi.
- `P1 = P2(...)`: Mengambil separuh spektrum positif (sesuai teorema Nyquist).
- `k = (0:(N/2)) / (N * dx)`: Mengonstruksi sumbu koordinat frekuensi spasial ($k$).

---

### 6. Contoh Hasil yang Diharapkan
- **Profil Spasial:** Menampilkan grafik gelombang halus (anomali regional) bermuatan riak frekuensi tinggi (anomali dangkal/noise).
- **Spektrum Daya:** Grafik linier meluruh ($\ln(P)$ vs $k$). Bagian curam di frekuensi rendah menandakan respon dalam, sedangkan bagian landai di frekuensi tinggi menandakan respon dangkal.

---

### 7. Penjelasan Grafik FFT
- **Frekuensi Rendah ($k \approx 0 - 0.005\text{ m}^{-1}$):** Nilai $\ln(P)$ sangat tinggi, menunjukkan keberadaan respon geologis regional skala besar (struktur dalam).
- **Frekuensi Sedang–Tinggi ($k > 0.01\text{ m}^{-1}$):** Amplitudo meluruh, merepresentasikan respon tubuh anomali lokal/dangkal serta efek *noise*.

---

### 8. Interpretasi Hasil dalam Konteks Geofisika
Menggunakan pendekatan Spector & Grant (1970), kedalaman rata-rata sumber anomali ($h$) berbanding lurus dengan kemiringan (*slope*, $m$) dari grafik $\ln(P)$ terhadap $k$:
$$h = -\frac{m}{4\pi}$$
Dengan mengukur kemiringan garis regresi pada bagian frekuensi rendah, geofisikawan dapat mengestimasi kedalaman basemen atau kedalaman struktur intrusi batuan tanpa harus melakukan pemboran.

---

### 9. Kemungkinan Error dan Cara Mengatasinya
1. **Error `Index exceeds array bounds`:** Terjadi jika `dx` atau indeks matriks melampaui jumlah elemen `N`.
   *Solusi:* Selalu periksa dimensi matriks dengan `size(x)` atau `length(x)`.
2. **Spectral Leakage:** Terjadi jika data tidak periodik atau Ujung awal dan akhir sinyal terputus tajam.
   *Solusi:* Terapkan *tapering window* (seperti `hann()` atau `hamming()`) sebelum `fft()`.

---

### 10. Kesimpulan Singkat
Analisis spektral FFT dengan MATLAB memudahkan pemisahan anomali regional (dalam) dan residual (dangkal) pada data medan potensial secara kuantitatif, seraya memberikan estimasi kedalaman sumber anomali bumi secara efisien.

# Output Prompt 3 — Respon Dosen Seismik Refraksi (Metode Seismik Refraksi)

---

### 1. Tujuan Pembelajaran
Setelah mengikuti materi ini, mahasiswa mampu:
- Menjelaskan mekanisme propagasi gelombang seismik refraksi di bawah permukaan bumi.
- Memahami dan menurunkan persamaan Hukum Snellius dan Sudut Kritis (*Critical Angle*).
- Membaca dan menganalisis Kurva Waktu Tempuh (*Travel-Time Curve*).
- Menghitung kecepatan lapisan ($v_1, v_2$) dan kedalaman batas lapisan (*interface depth*, $h$).

---

### 2. Konsep Dasar Seismik Refraksi
Metode seismik refraksi memanfaatkan pembiasan kritis gelombang seismik yang menjalar sepanjang batas lapisan batuan (*interface*) di bawah permukaan. Metode ini mensyaratkan nilai kecepatan lapisan bawah harus lebih besar daripada kecepatan lapisan atas ($v_2 > v_1$).

---

### 3. Istilah-Istilah Penting
1. **Source (Sumber Seismik):** Titik tempat pelepasan energi (pukulan godam/dinamit).
2. **Geophone (Receiver):** Sensor penangkap getaran tanah.
3. **Direct Wave (Gelombang Langsung):** Gelombang yang merambat lurus di lapisan 1 dari source ke geophone.
4. **Refracted Wave / Head Wave (Gelombang Refraksi):** Gelombang yang menjalar sepanjang batas lapisan dengan kecepatan $v_2$ dan memancarkan kembali energi ke permukaan.
5. **Intercept Time ($t_i$):** Waktu tempuh gelombang refraksi jika diekstrapolasikan mundur ke jarak $x = 0$.
6. **Crossover Distance ($x_{cross}$):** Jarak offset di mana gelombang refraksi menyusul dan tiba lebih awal daripada gelombang langsung.

---

### 4. Prinsip Fisika yang Digunakan
- **Prinsip Huygens:** Setiap titik pada muka gelombang bertindak sebagai sumber gelombang sekuler baru.
- **Prinsip Fermat:** Gelombang seismik menempuh lintasan dengan waktu paling minimum.

---

### 5. Hukum Snellius
Persamaan umum pembiasan gelombang:
$$\frac{\sin i_1}{v_1} = \frac{\sin i_2}{v_2}$$
Saat terjadi pembiasan kritis, sudut bias $i_2 = 90^\circ$ ($\sin 90^\circ = 1$), sehingga sudut datang $i_1$ menjadi **Sudut Kritis ($i_c$)**:
$$\sin i_c = \frac{v_1}{v_2}$$

---

### 6. Rumus-Rumus Penting (Model 2 Lapisan Datar)

1. **Waktu Tempuh Gelombang Langsung:**
   $$t_{\text{direct}} = \frac{x}{v_1}$$

2. **Waktu Tempuh Gelombang Refraksi:**
   $$t_{\text{refracted}} = \frac{x}{v_2} + t_i$$

3. **Waktu Intercept ($t_i$):**
   $$t_i = \frac{2h \sqrt{v_2^2 - v_1^2}}{v_1 v_2}$$

4. **Kedalaman Lapisan ($h$):**
   $$h = \frac{t_i}{2} \frac{v_1 v_2}{\sqrt{v_2^2 - v_1^2}}$$

---

### 7. Penurunan Rumus Kedalaman Lapisan ($h$)
1. **Lintasan total gelombang refraksi ($S \to A \to B \to G$):**
   - Lintasan turun ($SA$): $SA = \frac{h}{\cos i_c}$
   - Lintasan sepanjang interface ($AB$): $AB = x - 2h \tan i_c$
   - Lintasan naik ($BG$): $BG = \frac{h}{\cos i_c}$

2. **Waktu tempuh total ($t$):**
   $$t = \frac{SA}{v_1} + \frac{AB}{v_2} + \frac{BG}{v_1} = \frac{2h}{v_1 \cos i_c} + \frac{x - 2h \tan i_c}{v_2}$$

3. Subtitusi $\tan i_c = \frac{\sin i_c}{\cos i_c}$ dan $\sin i_c = \frac{v_1}{v_2}$:
   $$t = \frac{x}{v_2} + \frac{2h}{v_1 \cos i_c} \left(1 - \frac{v_1^2}{v_2^2}\right)$$

4. Karena $\cos i_c = \sqrt{1 - \sin^2 i_c} = \sqrt{1 - \frac{v_1^2}{v_2^2}} = \frac{\sqrt{v_2^2 - v_1^2}}{v_2}$, maka disederhanakan menjadi:
   $$t = \frac{x}{v_2} + \frac{2h \sqrt{v_2^2 - v_1^2}}{v_1 v_2}$$
   Suku kedua inilah yang disebut **$t_i$ (Intercept Time)**!

---

### 8. Contoh Perhitungan Sederhana
**Soal:** Diketahui data seismik refraksi 2 lapisan:
- Kemiringan garis gelombang langsung pada kurva travel time memberikan $v_1 = 500\text{ m/s}$.
- Kemiringan garis gelombang refraksi memberikan $v_2 = 2000\text{ m/s}$.
- Waktu intercept diperoleh $t_i = 0.04\text{ detik}$.
Hitung sudut kritis $i_c$ dan kedalaman lapisan pertama $h$.

**Pembahasan Langkah demi Langkah:**
1. *Hitung Sudut Kritis ($i_c$):*
   $$\sin i_c = \frac{v_1}{v_2} = \frac{500}{2000} = 0.25 \implies i_c = \arcsin(0.25) \approx 14.48^\circ$$

2. *Hitung Kedalaman Lapisan ($h$):*
   $$h = \frac{t_i}{2} \frac{v_1 v_2}{\sqrt{v_2^2 - v_1^2}}$$
   $$h = \frac{0.04}{2} \frac{(500)(2000)}{\sqrt{2000^2 - 500^2}} = 0.02 \times \frac{1,000,000}{\sqrt{4,000,000 - 250,000}}$$
   $$h = 0.02 \times \frac{1,000,000}{\sqrt{3,750,000}} = 0.02 \times \frac{1,000,000}{1936.49} \approx 10.33\text{ meter}$$

---

### 9. Contoh Soal Tingkat Kuliah

   $$v_1 = \frac{x}{t} = \frac{10\text{ m}}{0.02\text{ s}} = 500\text{ m/s}$$

# Output Prompt 4 — Respon Tutor Geofisika Komputasi (Pengolahan Data Geofisika MATLAB Lanjutan)

---

### 1. Analisis Struktur Data
Data geofisika yang diolah adalah sinyal rekaman spasial/waktu. Kode dirancang kuat (*robust*) untuk mengantisipasi data yang mengandung *missing values* (NaN) atau kesalahan format deret.

---

### 2. Penjelasan Metode FFT
FFT mengubah representasi sinyal dari domain spasial ($x$) ke domain bilangan gelombang ($k$). Komponen daya spektral mendiskriminasikan efek dangkal (frekuensi tinggi) dari anomali regional dalam (frekuensi rendah).

---

### 3. Langkah Pengolahan
1. **Pemeriksaan & Pembersihan Data:** Validasi input, pembuangan NaN, dan detrending.
2. **Estimasi Sampling Spasial:** Menghitung $\Delta x$ dan Nyquist Limit.
3. **Kalkulasi FFT & Power Spectrum:** Menggunakan fungsi `fft()` dan konversi ke skala logaritmik.
4. **Modul Handling Error:** Menangkap kesalahan dimensi matriks dengan blok `try-catch`.

---

### 4. Kode MATLAB Lengkap

```matlab
% =========================================================================
% SCRIPT MATLAB: ROBUST GEOFISIKA FFT PROCESSING & ERROR HANDLING
% =========================================================================
clc; clear; close all;

try
    %% 1. READ & VALIDATE DATA
    % Contoh data sintetis terstruktur
    x = linspace(0, 5000, 501); % Lintasan 5 km, 501 titik
    dx = x(2) - x(1);
    
    % Sinyal Geofisika Sintetis: Regional + Anomali Lokal
    signal_clean = 40*exp(-x/1500) + 10*sin(2*pi*x/200);
    noise = 1.5 * randn(size(x));
    data_geofisika = signal_clean + noise;
    
    % Validasi Ukuran Array
    if length(x) ~= length(data_geofisika)
        error('Ukuran matriks posisi x dan data geofisika tidak cocok!');
    end
    
    %% 2. PREPROCESSING DATA
    data_clean = detrend(data_geofisika - mean(data_geofisika));
    N = length(data_clean);
    
    %% 3. FFT COMPUTATION
    Y = fft(data_clean);
    P2 = abs(Y/N);
    P1 = P2(1:floor(N/2)+1);
    P1(2:end-1) = 2*P1(2:end-1);
    
    k = (0:(N/2)) / (N*dx); % Spatial frequency (cycle/m)
    power_spectrum = P1.^2;
    
    %% 4. PLOTTING
    figure('Color', 'w');
    subplot(2,1,1);
    plot(x, data_geofisika, 'k', 'LineWidth', 1); hold on;
    plot(x, data_clean, 'r--', 'LineWidth', 1.2);
    title('Data Profil Anomali Geofisika');
    xlabel('Jarak (m)'); ylabel('Amplitudo Anomali');
    legend('Data Mentah', 'Data Preprocessed'); grid on;
    
    subplot(2,1,2);
    semilogy(k, power_spectrum, 'b-o', 'LineWidth', 1.5, 'MarkerSize', 3);
    title('Power Spectrum FFT (Skala Logaritmik)');
    xlabel('Frekuensi Spasial k (cycle/m)'); ylabel('Power Spectrum');
    grid on;
    
    disp('Pengolahan FFT Berhasil Tanpa Error!');

catch ME
    fprintf('
 [ERROR TERDETEKSI]: %s
', ME.message);
    fprintf('Petunjuk Perbaikan: Periksa kembali variabel input dan tipe data anda.
');
end
```

---

### 5. Penjelasan Kode Baris demi Baris

- `detrend(...)`: Menghilangkan garis tren linier dan nilai rata-rata sinyal.
- `semilogy(...)`: Plot sumbu Y dalam skala logaritmik ($10^y$) untuk memperjelas perbedaan magnitudo spektrum daya.

---

### 6. Contoh Hasil yang Diharapkan
Script menghasilkan dua subplot: Grafik atas berupa sinyal anomali spasial dan Grafik bawah berupa kurva spektrum daya yang meluruh terhadap frekuensi spasial.

### 7. Penjelasan Grafik FFT

### 8. Interpretasi Hasil dalam Konteks Geofisika
