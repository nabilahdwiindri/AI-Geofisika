# PROMPT 1 ANALISIS GELOMBANG SEISMIK DAN GRAFIK WAKTU TEMPUH

## ROLE

Kamu adalah dosen Seismologi yang berpengalaman mengajar mahasiswa S1
Geofisika dan mampu menjelaskan konsep gelombang seismik dengan bahasa
sederhana.

## TASK

Bantu saya memahami hubungan antara jarak stasiun seismik, waktu tempuh
gelombang, dan kecepatan gelombang P. Buatlah grafik waktu tempuh
terhadap jarak dan jelaskan cara membaca grafik tersebut.

## CONTEXT

Saya mahasiswa semester 3 Program Studi Geofisika yang sedang
mempelajari gelombang seismik. Saya masih belajar memahami rumus dan
grafik, sehingga membutuhkan penjelasan bertahap dari konsep dasar.

## INPUT

Gunakan data sederhana berikut:

    Jarak (km)   Waktu tempuh (s)
  ------------ ------------------
            10                  2
            20                  4
            30                  6
            40                  8
            50                 10

Data tersebut merupakan data latihan sederhana untuk menggambarkan
hubungan jarak dan waktu tempuh gelombang.

## CONSTRAINTS

1.  Gunakan bahasa Indonesia yang sederhana dan mudah dipahami.
2.  Jelaskan pengertian gelombang P, jarak, waktu tempuh, dan kecepatan.
3.  Tuliskan rumus yang digunakan serta arti setiap simbol dan
    satuannya.
4.  Hitung kecepatan gelombang secara bertahap.
5.  Buat grafik dengan sumbu X sebagai jarak dan sumbu Y sebagai waktu
    tempuh.
6.  Berikan judul, label sumbu, satuan, titik data, dan garis yang
    sesuai pada grafik.
7.  Gunakan Python dan Matplotlib untuk menghasilkan grafik yang dapat
    ditampilkan. Jika grafik tidak dapat ditampilkan, berikan kode yang
    bisa dijalankan.
8.  Jangan langsung memberikan kesimpulan tanpa menunjukkan perhitungan.
9.  Jelaskan bahwa data ini adalah data latihan, bukan hasil pengukuran
    gempa sebenarnya.

## OUTPUT

Susun jawaban dengan urutan: 1. Tujuan pembelajaran. 2. Penjelasan
konsep dasar. 3. Rumus dan arti simbol. 4. Perhitungan kecepatan
gelombang P. 5. Grafik waktu tempuh terhadap jarak. 6. Cara membaca dan
menginterpretasikan grafik. 7. Kesimpulan singkat. 8. Dua soal latihan
tanpa jawaban terlebih dahulu.

membimbing mahasiswa Geofisika dalam pengolahan data sederhana.

gravitasi pengamatan dan gravitasi acuan. Selanjutnya, buat grafik

    Jarak (m)   Gravitasi pengamatan (mGal)   Gravitasi acuan (mGal)

          200                       980,080                  980,000
          300                       980,150                  980,000
          400                       980,110                  980,000

**Anomali gravitasi = gravitasi pengamatan − gravitasi acuan.**

## CONSTRAINTS

2.  Jelaskan terlebih dahulu apa itu gravitasi pengamatan, gravitasi

5.  Buat grafik jarak terhadap anomali gravitasi menggunakan Python dan

6.  Beri label sumbu X, sumbu Y, judul, satuan, dan titik pengukuran.
7.  Jelaskan titik dengan anomali terbesar dan terkecil berdasarkan

## OUTPUT

Berikan jawaban dengan format: 1. Tujuan pembelajaran. 2. Konsep dasar
anomali gravitasi. 3. Rumus dan arti simbol. 4. Tabel perhitungan
anomali. 5. Grafik anomali gravitasi terhadap jarak. 6. Penjelasan pola
grafik. 7. Contoh interpretasi sederhana dan keterbatasannya. 8.
Kesimpulan. 9. Dua soal latihan beserta data baru tanpa kunci jawaban.

# PROMPT 3 MEMAHAMI MEDAN MAGNET DI SEKITAR KAWAT BERARUS

## ROLE

Kamu adalah dosen Fisika Medan yang menguasai konsep medan magnet dan
mampu menjelaskan perhitungan fisika untuk mahasiswa S1 Geofisika.

## TASK

Bantu saya memahami bagaimana kuat medan magnet di sekitar kawat lurus
panjang berubah terhadap jarak. Hitung nilai medan magnet pada beberapa
jarak dan buat grafik hubungan antara jarak dengan kuat medan magnet.

## CONTEXT

Saya mahasiswa semester 3 Geofisika yang sedang mempelajari medan
magnet. Saya membutuhkan contoh sederhana untuk memahami hubungan antara
arus listrik, jarak, dan medan magnet tanpa perhitungan yang terlalu
rumit.

## INPUT

Sebuah kawat lurus panjang dialiri arus listrik sebesar 5 A. Hitung
medan magnet pada jarak: - 0,01 m - 0,02 m - 0,03 m - 0,04 m - 0,05 m

Gunakan persamaan:

**B = μ₀I/(2πr)**

Dengan: - B = kuat medan magnet dalam tesla (T). - μ₀ = permeabilitas
vakum = 4π × 10⁻⁷ T·m/A. - I = arus listrik dalam ampere (A). - r =
jarak dari kawat dalam meter (m).

## CONSTRAINTS

1.  Gunakan bahasa Indonesia yang sederhana.
2.  Jelaskan konsep medan magnet sebelum melakukan perhitungan.
3.  Jelaskan arti setiap simbol dan satuan.
4.  Hitung nilai B untuk setiap jarak secara bertahap.
5.  Gunakan notasi ilmiah jika diperlukan.
6.  Buat tabel hasil perhitungan.
7.  Buat grafik jarak pada sumbu X dan kuat medan magnet pada sumbu Y
    menggunakan Python dan Matplotlib.
8.  Tambahkan judul, label sumbu, satuan, dan titik data.
9.  Jelaskan mengapa medan magnet melemah ketika jarak bertambah.
10. Nyatakan bahwa perhitungan menggunakan model kawat lurus panjang
    ideal dan mengabaikan pengaruh medan magnet lain.

## OUTPUT

Susun hasil dengan urutan: 1. Tujuan pembelajaran. 2. Konsep dasar medan
magnet. 3. Rumus dan arti simbol. 4. Contoh perhitungan pada satu jarak.
5. Tabel hasil untuk seluruh jarak. 6. Grafik kuat medan magnet terhadap
jarak. 7. Interpretasi grafik. 8. Kesimpulan. 9. Dua latihan soal tanpa
jawaban terlebih dahulu.

# PROMPT 4 MENENTUKAN KEDALAMAN LAPISAN DENGAN SEISMIK REFRAKSI

## ROLE

Kamu adalah dosen Metode Seismik yang berpengalaman mengajar seismik
refraksi dan interpretasi lapisan bawah permukaan.

## TASK

Bantu saya memahami cara memperkirakan kedalaman batas antara dua
lapisan bawah permukaan menggunakan metode seismik refraksi. Jelaskan
cara memperoleh kecepatan lapisan dari grafik waktu tempuh terhadap
jarak dan cara menggunakan waktu intersep untuk menghitung kedalaman.

## CONTEXT

Saya mahasiswa semester 3 Geofisika yang baru mempelajari seismik
refraksi. Saya masih membutuhkan bantuan dalam memahami grafik,
menentukan gradien garis, dan memasukkan nilai ke dalam rumus.

## INPUT

Gunakan data latihan dari model dua lapisan berikut: - Kecepatan lapisan
pertama: v₁ = 500 m/s. - Kecepatan lapisan kedua: v₂ = 1500 m/s. - Waktu
intersep gelombang refraksi: tᵢ = 0,20 s. - Asumsikan dua lapisan datar
dan homogen, dengan v₂ lebih besar daripada v₁.

Gunakan rumus kedalaman:

**h = (tᵢ/2) × (v₁v₂/√(v₂² − v₁²))**

## CONSTRAINTS

1.  Gunakan bahasa Indonesia yang sederhana dan jelaskan konsep terlebih
    dahulu.
2.  Jelaskan perbedaan gelombang langsung dan gelombang refraksi.
3.  Jelaskan arti kecepatan lapisan, waktu intersep, dan kedalaman batas
    lapisan.
4.  Tunjukkan rumus waktu tempuh gelombang langsung dan refraksi.
5.  Buat grafik waktu tempuh terhadap jarak dengan dua garis yang
    mewakili gelombang langsung dan gelombang refraksi.
6.  Gunakan Python dan Matplotlib untuk membuat grafik yang memiliki
    label sumbu, satuan, legenda, dan judul.
7.  Gunakan persamaan t_langsung = x/v₁ dan t_refraksi = x/v₂ + tᵢ untuk
    membuat grafik.
8.  Hitung kedalaman lapisan langkah demi langkah dengan satuan yang
    konsisten.
9.  Jelaskan asumsi dan keterbatasan model dua lapisan.
10. Jangan mengarang hasil pengukuran lapangan; gunakan data yang
    diberikan sebagai model latihan.

## OUTPUT

Berikan hasil dengan urutan: 1. Tujuan pembelajaran. 2. Konsep dasar
seismik refraksi. 3. Rumus waktu tempuh dan kedalaman. 4. Grafik waktu
tempuh kedua gelombang. 5. Penjelasan cara membaca grafik. 6.
Perhitungan kedalaman lapisan. 7. Interpretasi hasil dan keterbatasan.
8. Kesimpulan. 9. Satu latihan soal baru tanpa jawaban.

# PROMPT 5 MEMAHAMI GELOMBANG SEISMIK MELALUI GRAFIK SINUSOIDAL

## ROLE

Kamu adalah dosen Fisika Gelombang yang ahli menjelaskan karakteristik
gelombang kepada mahasiswa S1 Geofisika.

## TASK

Bantu saya memahami amplitudo, periode, frekuensi, panjang gelombang,
dan kecepatan rambat gelombang melalui contoh gelombang sinusoidal. Buat
grafik gelombang dan tunjukkan bagian-bagian pentingnya.

## CONTEXT

Saya mahasiswa semester 3 Geofisika yang sedang mempelajari gelombang
mekanik dan gelombang seismik. Saya lebih mudah memahami materi jika
rumus disertai gambar atau grafik dan contoh angka yang sederhana.

## INPUT

Gunakan contoh gelombang seismik dengan parameter: - Amplitudo A = 2
cm. - Frekuensi f = 5 Hz. - Panjang gelombang λ = 100 m. - Fase awal φ =
0.

Gunakan persamaan:

**u(x,t) = A sin(kx − ωt)**

Dengan: - k = 2π/λ. - ω = 2πf. - v = fλ.

Untuk grafik pertama, gambarkan simpangan u terhadap posisi x pada saat
t = 0. Untuk grafik kedua, gambarkan simpangan u terhadap waktu t pada
posisi x = 0.

## CONSTRAINTS

1.  Gunakan bahasa Indonesia yang mudah dipahami.
2.  Jelaskan setiap parameter gelombang sebelum menggunakan rumus.
3.  Jelaskan arti simbol dan satuan.
4.  Hitung bilangan gelombang k, frekuensi sudut ω, periode T, dan
    kecepatan rambat v.
5.  Buat dua grafik menggunakan Python dan Matplotlib.
6.  Grafik pertama harus menunjukkan simpangan terhadap posisi,
    sedangkan grafik kedua menunjukkan simpangan terhadap waktu.
7.  Cantumkan judul, label sumbu, satuan, dan keterangan parameter.
8.  Tunjukkan amplitudo dan panjang gelombang pada grafik pertama serta
    periode pada grafik kedua.
9.  Jelaskan perbedaan grafik ruang dan grafik waktu.
10. Jangan menyamakan amplitudo dengan kecepatan atau frekuensi.
11. Gunakan parameter sebagai contoh matematis, bukan data rekaman gempa
    sebenarnya.

## OUTPUT

Susun hasil dengan urutan: 1. Tujuan pembelajaran. 2. Konsep dasar
gelombang. 3. Penjelasan parameter gelombang. 4. Rumus dan perhitungan.
5. Grafik simpangan terhadap posisi. 6. Grafik simpangan terhadap waktu.
7. Cara membaca kedua grafik. 8. Hubungan gelombang dengan seismologi.
9. Kesimpulan. 10. Dua latihan soal tanpa jawaban terlebih dahulu.

Kelima prompt ini menerapkan struktur prompt yang sistematis. **ROLE**
menetapkan peran AI, **TASK** menjelaskan tugas, **CONTEXT** memberikan
latar belakang, **INPUT** menyediakan data, **CONSTRAINTS** menetapkan
aturan pengerjaan, dan **OUTPUT** menentukan susunan jawaban yang
diharapkan.

Dengan struktur tersebut, instruksi menjadi lebih spesifik sehingga AI
diharapkan memberikan jawaban yang relevan, bertahap, dan mudah
dipahami. Hasil dari AI tetap perlu diperiksa kembali, terutama untuk
memastikan ketepatan perhitungan, satuan, grafik, dan interpretasi
geofisika.
